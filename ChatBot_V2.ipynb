```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Cargar el tokenizador y el modelo GPT-2
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

# Entrada del usuario
input_text = "Hola, ¿cómo estás?"

# Tokenizar la entrada
input_ids = tokenizer.encode(input_text, return_tensors="pt")

# Generar la respuesta
output = model.generate(input_ids, max_length=50, do_sample=True, temperature=0.7)

# Decodificar y mostrar la salida
response = tokenizer.decode(output[0], skip_special_tokens=True)
print(response)
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Cargar el modelo y el tokenizador GPT-2
model_name = 'gpt2'  # Puedes usar 'gpt2-medium', 'gpt2-large', o 'gpt2-xl' según el tamaño que prefieras
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# Función para generar texto
def generate_text(prompt, max_length=100):
    # Tokenizar el prompt
    inputs = tokenizer.encode(prompt, return_tensors='pt')

    # Generar texto con muestreo
    outputs = model.generate(inputs,
                             max_length=max_length,
                             num_return_sequences=1,
                             do_sample=True,  # Habilitar muestreo
                             no_repeat_ngram_size=2,
                             top_k=50,
                             top_p=0.95,
                             temperature=0.7,
                             pad_token_id=tokenizer.eos_token_id)  # Establecer el pad_token_id

    # Decodificar el texto generado
    text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return text

# Interacción con el usuario
print("¡Hola! Soy GPT-2. Puedes hacerme preguntas o darme prompts para generar texto. Para salir, escribe 'salir'.")

while True:
    user_input = input("\nPregunta o prompt: ")

    if user_input.lower() == 'salir':
        print("¡Hasta luego!")
        break

    # Generar respuesta
    response = generate_text(user_input)

    print("\nRespuesta generada:")
    print(response)
```
